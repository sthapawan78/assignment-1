Add

 using nested

: 85+ →

,

. Print the overall pass rate as a percentage by taking the mean of

 cast to

. Deal with the impossible marks values you found in C2, and say in a comment
what you did with them and why33

```python
# Marks outside 0–100 are physically impossible.
# Convert those values to NaN, then remove those invalid records before
# calculating pass rate and grades. This prevents impossible scores from distorting the derived results.
```

```python
invalid_mark_mask = ~df["marks"].between(0, 100)
invalid_mark_count = invalid_mark_mask.sum()
```

```python
print("Impossible marks found:", invalid_mark_count)
print("Impossible mark values:", df.loc[invalid_mark_mask, "marks"].tolist())
```

```python
df.loc[invalid_mark_mask, "marks"] = np.nan
df = df.dropna(subset=["marks"]).copy()
```

```python
# Pass/fail.
df["passed"] = df["marks"] >= 40
```

```python
# Nested np.where for the required grade bands:
# 85+ = A, 70–84 = B, 40–69 = C, below 40 = F.
df["grade"] = np.where(
    df["marks"] >= 85,
    "A",
    np.where(
        df["marks"] >= 70,
        "B",
        np.where(df["marks"] >= 40, "C", "F")
    )
)
```

```python
pass_rate = df["passed"].astype(int).mean() * 100
```

```python
print("\nOverall pass rate: {:.2f}%".format(pass_rate))
print("\nGrade counts:")
print(df["grade"].value_counts())
```

## C11 — Types and memory (3 pts)
Convert 
subject
 and 
city
 to the 
category
 dtype. Print 
memory_usage(deep=True)
 for those two
columns before and after, in bytes, and the percentage saved. In a comment, explain what makes 
category
cheaper, and when it would not help

```python
# Measure memory before converting to category.
memory_before = df[["subject", "city"]].memory_usage(deep=True)
```

```python
print("Memory before category conversion (bytes):")
print(memory_before)
```

```python
# Convert repeated text columns to category.
df["subject"] = df["subject"].astype("category")
df["city"] = df["city"].astype("category")
```

```python
memory_after = df[["subject", "city"]].memory_usage(deep=True)
```

```python
print("\nMemory after category conversion (bytes):")
print(memory_after)
```

```python
before_total = memory_before.sum()
after_total = memory_after.sum()
saved_bytes = before_total - after_total
saved_percent = (saved_bytes / before_total) * 100
```

```python
print("\nTotal before:", before_total, "bytes")
print("Total after:", after_total, "bytes")
print("Bytes saved:", saved_bytes)
print("Percentage saved: {:.2f}%".format(saved_percent))
```

```python
# category is cheaper when a column contains many repeated values because
# Pandas stores the unique values once and stores compact category codes.
# It may not help (and can sometimes use more memory) when most values are unique or the column is very small.
```

## C12 — Aggregation and a summary table (3 pts)
On the cleaned data, produce: (a) mean, max and count of 
mean

 by

 and

; (c) a

 per subject using 
groupby().agg() 
; (b)

 of mean marks with cities as rows and subjects as

```python
# (a) Mean, max and count of marks by city and subject.
summary_city_subject = (
    df.groupby(["city", "subject"], observed=True)["marks"]
      .agg(["mean", "max", "count"])
      .reset_index()
      .sort_values(["city", "subject"])
)
```

```python
print("(a) Mean, max and count of marks by city and subject:")
display(summary_city_subject)
```

```python
OUTPUT_DIR = "outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
# Export (a) to outputs/summary.csv.
summary_path = os.path.join(OUTPUT_DIR, "summary.csv")
summary_city_subject.to_csv(summary_path, index=False)

print("\nSaved:", summary_path)
```

```python
# (b) Mean marks per subject using groupby().agg().
summary_subject = (
    df.groupby("subject", observed=True)["marks"]
      .agg(mean_marks="mean", max_marks="max", count="count")
      .reset_index()
)
```

```python
print(" Mean/max/count marks per subject:")
display(summary_subject)
```

```python
# (c) Pivot table of mean marks: cities as rows, subjects as columns.
pivot_mean = pd.pivot_table(
    df,
    values="marks",
    index="city",
    columns="subject",
    aggfunc="mean",
    observed=True
)
```

```python
print("(c) Pivot table — mean marks:")
display(pivot_mean)
```

```python
# (d) Mean marks per exam_month, sorted chronologically.
month_summary = (
    df.groupby("exam_month", observed=True)["marks"]
      .mean()
      .sort_index()
)
```

```python
print("(d) Mean marks per exam month:")
display(month_summary)
```

```python
if len(month_summary) >= 2:
    trend_direction = "increases" if month_summary.iloc[-1] > month_summary.iloc[0] else "decreases" if month_summary.iloc[-1] < month_summary.iloc[0] else "stays roughly the same"
    print(
        f"Month trend: mean marks {trend_direction} from the first exam month "
        f"({month_summary.iloc[0]:.2f}) to the last exam month ({month_summary.iloc[-1]:.2f})."
    )
else:
    print("Month trend: there are not enough distinct exam months to describe a trend.")
```
