## B5 — Vectorization vs. a Python loop (4 pts)
Convert temps to Fahrenheit two ways: with an explicit loop (the only loop allowed in Part B) and with one vectorized expression. Confirm they agree with np.allclose. Then time both on an array of 1,000,000 values using time.perf_counter(), and print both timings plus the speedup as a multiple (~Nx faster ). In a comment, explain where the loop's extra time actually goes

```python
fahrenheit_loop=np.empty_like(temps)
for i in range(temps.shape[0]):
    for j in range(temps.shape[1]):
        fahrenheit_loop[i, j] = (temps[i, j] * 9 / 5) + 32
```

```python
# Vectorized expression
fahrenheit_vectorized = (temps * 9 / 5) + 32
```

```python
print("B5a — Loop and vectorized results agree:",
      np.allclose(fahrenheit_loop, fahrenheit_vectorized, equal_nan=True))
```

```python
# Timing on 1,000,000 values
large_values = np.random.randint(14, 36, size=1_000_000).astype(float)
```

```python
start_loop = __import__("time").perf_counter()

large_fahrenheit_loop = np.empty_like(large_values)

for i in range(large_values.size):
    large_fahrenheit_loop[i] = (large_values[i] * 9 / 5) + 32

loop_time = __import__("time").perf_counter() - start_loop
```

```python
start_vectorized = __import__("time").perf_counter()

large_fahrenheit_vectorized = (large_values * 9 / 5) + 32

vectorized_time = __import__("time").perf_counter() - start_vectorized
```

```python
speedup = loop_time / vectorized_time
```

```python
print("B5b — Loop timing:", loop_time, "seconds")
print("B5c — Vectorized timing:", vectorized_time, "seconds")
print("B5d — Vectorization speedup: ~", speedup, "x")
```

```python
# The loop spends extra time repeatedly executing Python-level iteration, indexing, and assignment, while NumPy performs the arithmetic in optimized compiled code over the whole array.
```

```python
quantities = np.array([2, 3, 1])
prices = np.array([120, 90, 250])

print("B7a — quantities * prices:", quantities * prices)
print("B7a — quantities @ prices:", quantities @ prices)

# * performs element-by-element multiplication and returns an array.
# @ performs dot/matrix multiplication and returns the total weighted sum.
```

```python
features = np.column_stack([
    np.nanmean(temps, axis=0),
    np.nanmean(humidity, axis=0)
])

weights = np.array([0.6, -0.1])

comfort_scores = features @ weights

print("B7b — features shape:", features.shape)
print("B7b — weights shape:", weights.shape)
print("B7b — comfort scores shape:", comfort_scores.shape)
print("B7b — comfort scores:", comfort_scores)
```

## B9 — Reproducible randomness and a train/test split (3 pts)
Flatten 
temps
 to 70 samples. With 
np.random.seed(42) 
, shuffle an index array and split it 80/20 into train
and test index sets; print both shapes, confirm they sum to 70, and confirm they share no index. Re-seed with
42, repeat, and assert the split is identical. In a comment, explain why you shuffle before splitting, and why
the seed matters when you share results

```python
flat_temps = temps.flatten()
```

```python
np.random.seed(42)

indices = np.arange(flat_temps.size)
np.random.shuffle(indices)

split_point = int(0.8 * len(indices))

train_indices = indices[:split_point]
test_indices = indices[split_point:]
```

```python
print("B9a — Train index shape:", train_indices.shape)
print("B9b — Test index shape:", test_indices.shape)
print("B9c — Total samples:", train_indices.size + test_indices.size)
print("B9d — Total is 70:", train_indices.size + test_indices.size == 70)
print("B9e — Train/test share no index:",
      np.intersect1d(train_indices, test_indices).size == 0)
```

```python
# Repeat with the same seed.
np.random.seed(42)

indices_repeat = np.arange(flat_temps.size)
np.random.shuffle(indices_repeat)

split_point_repeat = int(0.8 * len(indices_repeat))

train_indices_repeat = indices_repeat[:split_point_repeat]
test_indices_repeat = indices_repeat[split_point_repeat:]
```

```python
print("B9f — Split is identical after reseeding:",
      np.array_equal(train_indices, train_indices_repeat)
      and np.array_equal(test_indices, test_indices_repeat))

assert np.array_equal(train_indices, train_indices_repeat)
assert np.array_equal(test_indices, test_indices_repeat)
```

```python
# Shuffling before splitting helps avoid depending on the original ordering of the data. The seed makes the random shuffle reproducible so others can recreate the same train/test split when checking or sharing results.
```

## B10 — np.where , argmax , argmin (3 pts)
(a) Label every reading "hot" / "mild" / "cold" (above 30, 22–30, below 22) with nested np.where , and print the label grid's shape. 
(b) Given scores = np.array([0.08, 0.62, 0.21, 0.09]) and labels=["rain", "clear", "cloudy", "fog"] , print the predicted label using argmax. In a comment, explain why a classifier needs argmax rather than max.

```python
temperature_labels = np.where(
    temps > 30,
    "hot",
    np.where(
        temps >= 22,
        "mild",
        "cold"
    )
)

print("B10a — Temperature label grid:")
print(temperature_labels)
print("B10a — Label grid shape:", temperature_labels.shape)
```

```python
scores = np.array([0.08, 0.62, 0.21, 0.09])
labels = np.array(["rain", "clear", "cloudy", "fog"])

predicted_index = np.argmax(scores)
predicted_label = labels[predicted_index]

print("B10b — Predicted label:", predicted_label)
print("B10b — Predicted index:", predicted_index)
```

```python
# argmax returns the position of the largest score, which lets us use that position to select the corresponding label. max() returns only the value, so it cannot directly tell us which label produced that value.
```
